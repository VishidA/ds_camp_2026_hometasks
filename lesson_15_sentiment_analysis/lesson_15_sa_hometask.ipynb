```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from nltk.corpus import stopwords
import nltk
from nltk.tokenize import RegexpTokenizer
```

```python
fn='data/rt-polarity.neg'

with open(fn, "r",encoding='utf-8', errors='ignore') as f:
    content = f.read()
texts_neg=  content.splitlines()
print ('len of texts_neg = {:,}'.format (len(texts_neg)))
for review in texts_neg[:5]:
    print ( '\n', review)
```

```python
fn='data/rt-polarity.pos'

with open(fn, "r",encoding='utf-8', errors='ignore') as f:
    content = f.read()
texts_pos=  content.splitlines()
print ('len of texts_pos = {:,}'.format (len(texts_pos)))
for review in texts_pos[:5]:
    print ('\n', review)
```

```python
def preproces_tokenize(text):
    tokenizer = RegexpTokenizer(r'\w+')
    return tokenizer.tokenize(text)

words_pos = [preproces_tokenize(review) for review in texts_pos]
words_neg = [preproces_tokenize(review) for review in texts_neg]

print(words_pos[:2])
print(words_neg[:2])
```

```python
nltk.download('stopwords')
```

```python
def remove_stop_words(words):
    stop_words = set(stopwords.words('english'))
    return [w for w in words if w not in stop_words]

words_pos_filtered = [remove_stop_words(words) for words in words_pos]
words_neg_filtered = [remove_stop_words(words) for words in words_neg]

print(words_neg_filtered[:2])
print(words_pos_filtered[:2])
```

```python
labels_pos = [1] * len(words_pos_filtered)
labels_neg = [0] * len(words_neg_filtered)

X = words_pos_filtered + words_neg_filtered
y = labels_pos + labels_neg
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
```

Task: Implement and evaluate at least 3 methods, reporting Accuracy, Precision, Recall, and F1-score

Methods to implement:
1. **NLTK Naive Bayes**
2. **TF-IDF + Logistic Regression**
3. **Word2Vec + Logistic Regression**

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from gensim.models import Word2Vec
import warnings
warnings.filterwarnings('ignore')
```

```python
def evaluate_and_print_metrics(y_test, predictions, method_name):

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(y_test, predictions, zero_division=0)
    recall = recall_score(y_test, predictions, zero_division=0)
    f1 = f1_score(y_test, predictions, zero_division=0)
    
    print(f"METHOD: {method_name}")
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-Score:  {f1:.4f}")
    
    return {'accuracy': accuracy, 'precision': precision, 'recall': recall, 'f1': f1}
```

```python
X_train_text = [' '.join(words) for words in X_train]
X_test_text = [' '.join(words) for words in X_test]

print(f"Training set size: {len(X_train)}")
print(f"Test set size: {len(X_test)}")
```

### Method 1: NLTK Naive Bayes Classifier
**Description**: Traditional bag-of-words approach with Naive Bayes classifier from NLTK library


```python
def document_features(words):
    return {word: True for word in words}

train_data = [(document_features(x), label) for x, label in zip(X_train, y_train)]
test_data = [(document_features(x), label) for x, label in zip(X_test, y_test)]

clf = nltk.NaiveBayesClassifier.train(train_data)

nltk_predictions = [clf.classify(document_features(x)) for x in X_test]

nltk_metrics = evaluate_and_print_metrics(y_test, nltk_predictions, "NLTK Naive Bayes Classifier")
nltk_accuracy = nltk_metrics['accuracy']
nltk_precision = nltk_metrics['precision']
nltk_recall = nltk_metrics['recall']
nltk_f1 = nltk_metrics['f1']
```

### Method 2: TF-IDF + Logistic Regression
**Description**: TF-IDF vectorization (numerical statistic reflecting word importance) combined with Logistic Regression. This is an effective traditional approach.


```python
tfidf_vectorizer = TfidfVectorizer(max_features=5000, min_df=2, max_df=0.8)
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train_text)
X_test_tfidf = tfidf_vectorizer.transform(X_test_text)

tfidf_lr_clf = LogisticRegression(max_iter=1000, random_state=42)
tfidf_lr_clf.fit(X_train_tfidf, y_train)

tfidf_predictions = tfidf_lr_clf.predict(X_test_tfidf)

tfidf_metrics = evaluate_and_print_metrics(y_test, tfidf_predictions, "TF-IDF + Logistic Regression")
tfidf_accuracy = tfidf_metrics['accuracy']
tfidf_precision = tfidf_metrics['precision']
tfidf_recall = tfidf_metrics['recall']
tfidf_f1 = tfidf_metrics['f1']
```

### Method 3: Word2Vec + Logistic Regression
**Description**: Word2Vec creates dense vector representations of words that capture semantic meaning. This modern approach learns word embeddings from the corpus and creates document vectors by averaging word vectors, then applies Logistic Regression for classification.

```python
w2v_model = Word2Vec(sentences=X_train, vector_size=100, window=5, min_count=2, workers=4)

def get_document_vector(words, model):
    vectors = [model.wv[word] for word in words if word in model.wv]
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    return np.mean(vectors, axis=0)

X_train_w2v = np.array([get_document_vector(words, w2v_model) for words in X_train])
X_test_w2v = np.array([get_document_vector(words, w2v_model) for words in X_test])

w2v_lr_clf = LogisticRegression(max_iter=1000, random_state=42)
w2v_lr_clf.fit(X_train_w2v, y_train)

w2v_predictions = w2v_lr_clf.predict(X_test_w2v)

w2v_metrics = evaluate_and_print_metrics(y_test, w2v_predictions, "Word2Vec + Logistic Regression")
w2v_accuracy = w2v_metrics['accuracy']
w2v_precision = w2v_metrics['precision']
w2v_recall = w2v_metrics['recall']
w2v_f1 = w2v_metrics['f1']
```

## Comparison of All Methods


```python
comparison_df = pd.DataFrame({
    'Method': [
        'NLTK Naive Bayes',
        'TF-IDF + Logistic Regression',
        'Word2Vec + Logistic Regression'
    ],
    'Accuracy': [nltk_accuracy, tfidf_accuracy, w2v_accuracy],
    'Precision': [nltk_precision, tfidf_precision, w2v_precision],
    'Recall': [nltk_recall, tfidf_recall, w2v_recall],
    'F1 Score': [nltk_f1, tfidf_f1, w2v_f1]
})

display_df = comparison_df.sort_values(by='F1 Score', ascending=False).reset_index(drop=True)
display_df
```

# Conclusion

I tested three different ways to teach a computer to recognize whether movie reviews are positive or negative. Each method works differently.

The first method is Naive Bayes from the NLTK library. It's the simplest approach. It just counts which words appear most often in positive and negative reviews, and then guesses what type a new review is. It works, but it's not very accurate.

The second method is TF-IDF with Logistic Regression. It's more complex than the first one. Here, the computer learns which words are most important for determining the mood of a review. Some words count more, some count less. This method shows better results than the first one.

The third method is Word2Vec with Logistic Regression. It's the most modern approach. Word2Vec first learns which words have similar meanings and how they relate to each other. Then it uses this information to classify reviews. Usually, this method gives the best results.

In general, as the methods become more sophisticated, they tend to understand the meaning of reviews better and make fewer mistakes.
